````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

R, L, K_t, K_e, J, b = 1.0, 0.01, 0.1, 0.1, 0.01, 0.001

def motor_dynamics(t, state, voltage, load_torque):
    current, omega = state
    return [(voltage - R*current - K_e*omega)/L,
            (K_t*current - b*omega - load_torque)/J]
class PIController:
    def __init__(self, kp, ki):
        self.kp, self.ki = kp, ki
        self.integral = 0.0
        self.d_term = 0.0
    def update(self, setpoint, measurement, dt):
        error = setpoint-measurement
        self.integral += error*dt
        self.p_term = self.kp*error
        self.i_term = self.ki*self.integral
        return self.p_term+self.i_term
def simulate(controller, target=lambda t: 40.0, load=lambda t: 0.0,
             duration=8.0, dt=0.001, noise=0.0, seed=42):
    # Each row describes the beginning of one control interval.
    count = round(duration/dt)
    if dt <= 0 or not np.isclose(count*dt, duration):
        raise ValueError("duration must be a positive integer multiple of dt")
    rng = np.random.default_rng(seed)
    state = np.array([0.0, 0.0])
    records = []
    for k in range(count):
        t = k*dt
        reference = target(t)
        measured = state[1] + noise*rng.normal()
        voltage = controller.update(reference, measured, dt)
        requested = getattr(controller, "requested", voltage)
        records.append([t, reference, state[0], state[1], measured,
                        requested, voltage, controller.p_term,
                        controller.i_term, controller.d_term])
        result = solve_ivp(motor_dynamics, (t, t+dt), state,
                          args=(voltage, load(t)), rtol=1e-7, atol=1e-9)
        if not result.success:
            raise RuntimeError(result.message)
        state = result.y[:, -1]
    keys = "time target current speed measured requested voltage p i d".split()
    return dict(zip(keys, np.asarray(records).T))

def plot_run(run, title):
    fig, axes = plt.subplots(3, 1, sharex=True, figsize=(10, 8), layout="constrained")
    axes[0].plot(run['time'], run['target'], '--', label='Target')
    axes[0].plot(run['time'], run['speed'], label='True speed')
    axes[0].set_ylabel('Speed (rad/s)')
    axes[1].plot(run['time'], run['requested'], label='Requested')
    axes[1].plot(run['time'], run['voltage'], '--', label='Applied')
    axes[1].set_ylabel('Voltage (V)')
    for key in ['p', 'i', 'd']:
        axes[2].plot(run['time'], run[key], label=key.upper())
    axes[2].set_ylabel('Contribution (V)')
    axes[2].set_xlabel('Time (s)')
    for ax in axes:
        ax.grid(alpha=.25); ax.legend()
    fig.suptitle(title)
    plt.show()
````

## 5.1 Estimate a slope from samples

A backward difference estimates the error derivative:

$$\dot e_k\approx\frac{e_k-e_{k-1}}{\Delta t},\qquad D_k=K_d\dot e_k.$$

For errors 10 then 8 over 0.1 s, the estimated slope is −20 rad/s². With $K_d=0.01$ V/(rad/s²), derivative voltage is −0.2 V. When a fixed target is approached, decreasing error produces a braking contribution.

### A derivative is sensitive to time and direction

A change from error 10 to 8 over 0.1 s produces a slope of −20 rad/s². The same change over 0.01 s produces −200 rad/s². Always use the actual elapsed controller interval in the denominator.

At a fixed target, error is target minus speed, so its derivative is negative angular acceleration. A rising speed therefore contributes a negative derivative command, tending to oppose rapid motion toward the target. This explains the damping intuition without implying that D is an independent speed target.

A derivative term has no persistent output when error is constant. It cannot replace integral action for maintaining the extra voltage required by an unknown constant load. Even a large constant error has zero derivative once it stops changing.

## 5.2 Add the third term

The controller now combines present error, accumulated error, and error slope:

$$V_k=K_p e_k+K_i I_{e,k}+K_d\frac{e_k-e_{k-1}}{\Delta t}.$$

The first update has no previous sample. We define its derivative as zero, rather than inventing a previous error. This suppresses only the initial derivative spike; later setpoint steps can still create a kick.

### History has an initialization problem

The first call contains a current error but no earlier error. Assuming the previous error was zero would create an artificial startup difference. We instead set the first derivative to zero and record the first error for use on the second call.

On every later call, calculate the difference before replacing the stored error. Replacing it first would subtract the new value from itself and make D permanently zero. This is a state-order bug that a hand-calculated two-sample test catches immediately.

The integral still updates on the first call. P, I, and D do not share the same initialization requirements. Preserve their separate contributions in the log so a large output can be traced to the correct term.

```python
class ThreeTermController:
    def __init__(self, kp, ki, kd):
        self.kp, self.ki, self.kd = kp, ki, kd
        self.integral = 0.0
        self.previous_error = None
    def update(self, setpoint, measurement, dt):
        error = setpoint-measurement
        derivative = 0.0 if self.previous_error is None else (error-self.previous_error)/dt
        self.integral += error*dt
        self.previous_error = error
        self.p_term, self.i_term, self.d_term = self.kp*error, self.ki*self.integral, self.kd*derivative
        return self.p_term+self.i_term+self.d_term

c=ThreeTermController(.4,.5,.01)
c.update(10,0,.1);assert c.d_term==0
c.update(8,0,.1);np.testing.assert_allclose(c.d_term,-.2)
```

The arithmetic is identical to the hand calculation, while the full output also contains P and I.

</details>

## 5.3 Compare PI and PID

For a constant reference, derivative-on-error is negative derivative-on-speed. It can add damping, but excessive derivative gain also creates large command changes. Evaluate the whole transient rather than assuming a lower peak alone means improvement.

A useful way to understand damping is to temporarily set integral gain to zero. For ideal continuous PD control and a constant target, eliminating current gives the homogeneous equation for speed deviation $\delta\omega=\omega-\omega_{ss}$ from equilibrium

$$LJ\ddot{\delta\omega}+(Lb+RJ+K_tK_d)\dot{\delta\omega}+(Rb+K_tK_e+K_tK_p)\delta\omega=0.$$

The derivative term increases the coefficient multiplying speed's rate of change. In this second-order case that can reduce oscillation. Adding integral action changes the closed-loop dynamics again, so the same intuition does not guarantee less overshoot in every PID experiment.

We will use two comparisons: the original PI/PID case and a deliberately aggressive P/PD case. The latter requests 400 V at startup with gain 10, so it is strictly an ideal-actuator teaching example, not a hardware tuning recommendation.


### Read both comparisons rather than seeking one verdict

In the moderate-gain PI/PID comparison, adding derivative action raises the observed peak slightly. It slows part of the response, changing how much integral correction has accumulated. A braking contribution at one instant can coexist with a larger peak later because the terms interact over time.

The aggressive P/PD comparison isolates a clearer damping example by removing integral action. Here D lowers the peak. Its initial command demand is far beyond the later hardware bound, so the example demonstrates dynamics rather than recommending those gains.

When explaining the curves, identify the gain tuple, target, sampling interval, and voltage assumptions. “Derivative reduces overshoot” without those qualifications would overgeneralize the experiment.

<details>
<summary>Deeper derivation: eliminate current in the PD case</summary>

For a constant load, the mechanical equation gives $K_t i=J\dot\omega+b\omega+\tau_L$. Differentiating it gives $K_t\dot i=J\ddot\omega+b\dot\omega$. Multiply the electrical equation by $K_t$ and substitute both expressions:

$$LJ\ddot\omega+(Lb+RJ)\dot\omega+(Rb+K_tK_e)\omega=K_tV-R\tau_L.$$

For constant reference and PD control, $V=K_p(r-\omega)-K_d\dot\omega$. Substitute that expression and collect terms. The constant forcing on the right is balanced at equilibrium. Subtract the equilibrium equation, leaving the homogeneous equation for $\delta\omega$ shown above. The derivative gain increases the coefficient of the first derivative, while proportional gain changes the coefficient of speed deviation. This derivation assumes continuous-time PD and constant load; our implementation is sampled, so numerical experiments remain necessary.

</details>

## 5.4 Expose kick and noise

A step in target is also a step in error. Dividing that change by a short sample interval creates a large derivative command. A 20 rad/s setpoint jump over 1 ms contributes approximately 200 V when $K_d=0.01$.

Noise causes a related problem: differencing successive noisy measurements amplifies rapid variation. We add noise only to the sensor measurement; the true physical state remains separate.

<div class="lesson-callout" style="border-left: 3px solid #b99a62; background-color: rgba(185, 154, 98, 0.08); padding: 12px 18px; margin: 16px 0;">

**Key distinction**

A noisy measured speed is not the true motor speed. Compare both and inspect the command. Lesson 9 will address kick with derivative-on-measurement and noise with filtering.

</div>

### Two different sources of rapid error changes

For derivative-on-error, subtract successive errors:

$$e_k-e_{k-1}=(r_k-r_{k-1})-(\omega_k-\omega_{k-1}).$$

The first difference is a change we deliberately commanded; the second is a measured response. At a setpoint step, the first can be large while the rotor has barely moved. This is why a 20 rad/s target step over 1 ms causes a derivative kick even with a perfect sensor.

Noise is different. Even with a constant target, successive measurement errors fluctuate. With independent noise of standard deviation $\sigma$, the noise-only sample difference has standard deviation $\sqrt{2}\sigma$. Its derivative contribution scales approximately as $K_d\sqrt{2}\sigma/\Delta t$ before closed-loop interactions. With our values this is about 2.83 V—substantial relative to a 6 V drive.

The later practical lesson treats the mechanisms separately: use measurement differentiation to avoid direct target kick, and filtering to reduce rapid measurement fluctuations. Neither changes the physical motor equations.

## What the experiment establishes

Derivative action can damp a fast response, but it can also worsen a particular PI transient and amplify discontinuities or measurement noise. You now have all three terms; next we will package their state and behavior into a reusable, testable controller.
