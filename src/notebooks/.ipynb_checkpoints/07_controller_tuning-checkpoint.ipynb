**Setup inventory:** `motor_dynamics` evaluates the two motor derivatives using SI units and state order `[current, omega]`. Where supplied, `simulate` advances the sampled controller and motor separately, and `plot_run` displays speed, voltage, and term contributions. Controller classes in setup were taught previously; their histories start fresh when you construct an instance. Run setup after restarting the kernel, then execute your exercise cells from top to bottom.

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

R, L, K_t, K_e, J, b = 1.0, 0.01, 0.1, 0.1, 0.01, 0.001

def motor_dynamics(t, state, voltage, load_torque):
    current, omega = state
    return [(voltage - R*current - K_e*omega)/L,
            (K_t*current - b*omega - load_torque)/J]
class PID:
    def __init__(self, kp, ki, kd):
        self.kp, self.ki, self.kd = kp, ki, kd
        self.reset()
    def reset(self):
        self.integral = 0.0
        self.previous_error = None
        self.p_term = self.i_term = self.d_term = 0.0
    def update(self, setpoint, measurement, dt):
        if not np.isfinite(dt) or dt <= 0:
            raise ValueError("dt must be finite and positive")
        error = setpoint-measurement
        slope = 0.0 if self.previous_error is None else (error-self.previous_error)/dt
        self.integral += error*dt
        self.previous_error = error
        self.p_term = self.kp*error
        self.i_term = self.ki*self.integral
        self.d_term = self.kd*slope
        return self.p_term+self.i_term+self.d_term
def simulate(controller, target=lambda t: 40.0, load=lambda t: 0.0,
             duration=8.0, dt=0.001, noise=0.0, seed=42):
    # Each row describes the beginning of one control interval.
    count = round(duration/dt)
    if dt <= 0 or not np.isclose(count*dt, duration):
        raise ValueError("duration must be a positive integer multiple of dt")
    rng = np.random.default_rng(seed)
    state = np.array([0.0, 0.0])
    records = []
    for k in range(count):
        t = k*dt
        reference = target(t)
        measured = state[1] + noise*rng.normal()
        voltage = controller.update(reference, measured, dt)
        requested = getattr(controller, "requested", voltage)
        records.append([t, reference, state[0], state[1], measured,
                        requested, voltage, controller.p_term,
                        controller.i_term, controller.d_term])
        result = solve_ivp(motor_dynamics, (t, t+dt), state,
                          args=(voltage, load(t)), rtol=1e-7, atol=1e-9)
        if not result.success:
            raise RuntimeError(result.message)
        state = result.y[:, -1]
    keys = "time target current speed measured requested voltage p i d".split()
    return dict(zip(keys, np.asarray(records).T))

def plot_run(run, title):
    fig, axes = plt.subplots(3, 1, sharex=True, figsize=(10, 8), layout="constrained")
    axes[0].plot(run['time'], run['target'], '--', label='Target')
    axes[0].plot(run['time'], run['speed'], label='True speed')
    axes[0].set_ylabel('Speed (rad/s)')
    axes[1].plot(run['time'], run['requested'], label='Requested')
    axes[1].plot(run['time'], run['voltage'], '--', label='Applied')
    axes[1].set_ylabel('Voltage (V)')
    for key in ['p', 'i', 'd']:
        axes[2].plot(run['time'], run[key], label=key.upper())
    axes[2].set_ylabel('Contribution (V)')
    axes[2].set_xlabel('Time (s)')
    for ax in axes:
        ax.grid(alpha=.25); ax.legend()
    fig.suptitle(title)
    plt.show()
````

## 7.1 Define performance before tuning

For a positive step from zero to a fixed target, **rise time** is the interval from first reaching 10% to first reaching 90%. **Overshoot** is the largest excess above target, expressed as a percentage. **Settling time** is the first sample after the last violation of a ±2% target band.

A finite simulation cannot establish that a response will stay settled forever. If it ends outside the band, report settling time as unavailable. Our final error is the mean error over the last 0.5 s: call it a finite-window estimate rather than assuming equilibrium.

### Work through a sampled response

Consider a target of 40 rad/s and this deliberately short table:

| Time (s) | Speed (rad/s) |
| --- | --- |
| 0.0 | 0 |
| 0.1 | 4 |
| 0.2 | 20 |
| 0.3 | 36 |
| 0.4 | 44 |
| 0.5 | 39.5 |
| 0.6 | 41 |
| 0.7 | 40.2 |
| 0.8 | 40 |

The 10% threshold is 4 rad/s and the 90% threshold is 36 rad/s, so sample-based rise time is $0.3-0.1=0.2$ s. Peak speed is 44, giving $(44-40)/40\times100=10\%$ overshoot.

The ±2% band is 39.2–40.8 rad/s. The response enters at 0.5 s, leaves at 0.6 s, and returns at 0.7 s. The observed settling time is therefore 0.7 s, not 0.5 s. The last sample outside the band is the critical one.

These are sample-based estimates without crossing interpolation. Finer samples can change them slightly. The definition is still useful if every candidate uses the same grid.

```python
def metrics(run, target=40.0, band=.02):
    t, y = run['time'], run['speed']
    def first(level):
        ids = np.flatnonzero(y >= level)
        return None if len(ids)==0 else float(t[ids[0]])
    t10, t90 = first(.1*target), first(.9*target)
    outside = np.flatnonzero(np.abs(y-target) > band*abs(target))
    settling = (float(t[0]) if len(outside)==0 else
                None if outside[-1]==len(t)-1 else float(t[outside[-1]+1]))
    tail = t >= t[-1]-.5
    return dict(rise=None if t10 is None or t90 is None else t90-t10,
                settling=settling, overshoot=max(0.0, float(y.max()/target-1))*100,
                final_error=float(target-y[tail].mean()),
                peak_voltage=float(np.max(np.abs(run['voltage']))))
```

These definitions are intentionally limited to positive, constant step targets starting at zero. Different experiments need different metric definitions.

</details>

## 7.2 Check the measurement code

Metrics can be wrong even when the controller is correct. Test a response that never reaches the target, and one that reaches it then ends outside the band. Both must avoid reporting a misleading settling time.

### Missing measurements are not zero measurements

If speed never reaches 90% of target, rise time is unavailable. Returning zero would incorrectly rank the slow response as exceptionally fast. If the last sample is outside the settling band, the observation does not establish settling.

A response may also end inside the band after only one sample. Our metric reports what was observed, not a proof that it will remain there indefinitely. Extend the simulation when that distinction matters.

Final error uses the mean over a specified window. Oscillation can average to zero while instantaneous errors remain large, so pair final error with a plot and settling criterion. Peak voltage is an effort metric, not a tracking metric; a candidate can score well on one while poorly on another.

## 7.3 Sweep one factor at a time

Begin with a baseline and change one parameter while holding the others fixed. Compare a proportional sweep at fixed integral and derivative gains, then an integral sweep at fixed proportional gain. Always reset controller state.

Report speed performance together with voltage demand. The ideal controller has no voltage limit, so a fast candidate may be infeasible for the later ±6 V actuator.

<div class="lesson-callout" style="border-left: 3px solid #b99a62; background-color: rgba(185, 154, 98, 0.08); padding: 12px 18px; margin: 16px 0;">

**Key distinction**

A gain is not good or bad in isolation. Its effect depends on the plant, sampling, target, load, and actuator constraints.

</div>

### Make the sweep interpretable

A proportional sweep holds Ki and Kd fixed while changing Kp. An integral sweep then holds Kp and Kd fixed. This isolates local effects; changing all gains together might improve the curve but would not explain which change mattered.

The printed table should use consistent units: seconds for rise and settling, percent for overshoot, rad/s for final error, and volts for peak demand. An unavailable value should remain visibly unavailable rather than being dropped from the report.

The baseline appears in both conceptual sweeps. It needs only one simulation because its inputs are identical. All candidates start with zero motor state and zero controller history.

A larger proportional gain can shorten initial rise while leaving a slower final correction controlled by the integral. This is why fastest rise and fastest settling need not choose the same candidate.

## 7.4 Select and challenge a candidate

For this exercise, require overshoot below 10%, final error magnitude below 0.1 rad/s, and observed settling. Among qualifying candidates choose the shortest settling time, then report peak voltage separately. This is an explicit ideal-actuator objective, not a hardware recommendation.

Check the selected response at half the sample interval. Later lessons will challenge it with disturbances and saturation.

### Selection is a constrained decision

First eliminate candidates that violate the stated criteria. Then rank the survivors by settling time. Ranking before filtering can choose a fast but unacceptable response. If no candidate qualifies, report that fact instead of silently relaxing the objectives.

For this experiment the selected gains are 0.4, 0.5, and 0. Their observed overshoot is below 1% and settling is about 0.785 s. The peak command is about 16 V, so the selection relies on the ideal actuator assumption. It is not a ready-to-use tuning for the later 6 V drive.

The half-time-step check challenges the numerical choice, while the following lessons challenge the operating conditions. Both are necessary forms of scrutiny, but neither alone establishes robust stability over all loads, noise levels, or parameter uncertainty.
