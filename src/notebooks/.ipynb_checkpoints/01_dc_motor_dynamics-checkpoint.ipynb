# 1. DC Motor Dynamics

Before we can control a motor, we need a model of how it behaves. In this section, we'll build a simulation that takes an applied voltage and predicts the motor's current, torque, and speed over time.

By the end, you'll have a motor model that later sections can use to test a controller. For now, voltage is an input we choose directly.

## From voltage to motion

A DC motor converts electrical energy into mechanical motion. Current flowing through its windings interacts with a magnetic field to produce **torque**: a twisting effort on the shaft. If that torque exceeds the opposing torque, the shaft accelerates.

```text
Applied voltage → Current → Motor torque → Acceleration → Speed
                     ↑                                      │
                     └──── Back EMF increases with speed ───┘
```

There is also an electrical effect in the other direction. A spinning motor generates a voltage called **back electromotive force**, or **back EMF**. During normal motoring, it opposes the applied voltage. As speed increases, less voltage is available to drive current through the windings.

This coupling explains why we must simulate current and speed together.

## What our model includes

We'll model a brushed DC motor with a constant magnetic field and fixed parameters. We'll include winding resistance and inductance, rotor inertia, and friction proportional to speed. We'll treat the supply as an ideal voltage source.

We won't model brush switching, heating, magnetic saturation, or static friction. This simplified model is enough to explore the relationship between voltage and speed, but its parameters are illustrative rather than a specification for a real motor.

Choose one direction of rotation as positive. Positive current produces positive motor torque. A positive external load torque in our equations acts against that positive direction.

## 1.1 Describe the motor's state

A **state** is a quantity we track as the simulation advances. Our motor has two state variables:

- **Current `i`**, measured in amperes (A).
- **Angular velocity `ω`**, measured in radians per second (rad/s). We'll call it `omega` in code.

Angular velocity describes how fast the shaft rotates. One complete turn is `2π` radians. To convert radians per second to revolutions per minute, multiply by `60 / (2π)`.

We don't need shaft angle for this speed-control tutorial. If we later wanted position, we could obtain it by integrating angular velocity.

The quantities below are **parameters**: they describe the motor and remain fixed during an experiment.

| Parameter | Meaning | Illustrative value | Unit |
| --- | --- | --- | --- |
| `R` | Winding resistance | 1.0 | Ω |
| `L` | Winding inductance | 0.01 | H |
| `K_t` | Torque produced per ampere | 0.1 | N·m/A |
| `K_e` | Back EMF produced per unit speed | 0.1 | V·s/rad |
| `J` | Rotational inertia | 0.01 | kg·m² |
| `b` | Viscous friction coefficient | 0.001 | N·m·s/rad |

Resistance limits current and dissipates energy as heat. Inductance resists changes in current. Inertia resists changes in speed. Viscous friction opposes rotation with a torque whose magnitude increases with speed.

For this ideal motor, `K_t` and `K_e` have equal numerical values when expressed in these SI units. They describe different relationships, so we keep their names separate.

## 1.2 How current changes

The applied voltage is balanced by three contributions: the voltage across the resistance, the voltage across the inductance, and the back EMF.

```text
voltage = R · i + L · di/dt + K_e · ω
```

Here, `di/dt` means the **rate of change of current**, in amperes per second. Rearranging gives the electrical equation we will implement:

```text
di/dt = (voltage − R · i − K_e · ω) / L
```

Read the numerator as the voltage left over after the resistive drop and back EMF. A positive remainder makes current increase; a negative remainder makes it decrease. When the remainder is zero, current is momentarily unchanged.

At startup, both current and speed are zero. The initial current slope is therefore `voltage / L`. With our values, it is **600 A/s**. This is a rate, not an instantaneous current of 600 A: inductance means current builds continuously from zero.

As current grows, the resistive drop grows. As the shaft accelerates, back EMF grows too. Both reduce the rate at which current increases.

## 1.3 How speed changes

Motor torque is proportional to current:

```text
motor_torque = K_t · i
```

Torque is not speed. Torque changes speed when it is not balanced by opposing torque. The rotational version of Newton's second law is:

```text
J · dω/dt = motor_torque − friction_torque − load_torque
```

Using our motor and friction relationships gives:

```text
dω/dt = (K_t · i − b · ω − load_torque) / J
```

`dω/dt` is **angular acceleration**, measured in rad/s². More net torque means more acceleration; more inertia means less acceleration for the same net torque.

At the exact starting instant, current is zero, so motor torque and acceleration are also zero in our unloaded experiment. Current starts increasing immediately, and the resulting torque then accelerates the shaft.

At steady speed, acceleration is zero because motor torque balances friction and any external load. A motor can therefore rotate at a constant speed while still producing torque and drawing current.

## 1.4 Turn the equations into a simulation

We now know how quickly current and speed change for any given state. A numerical integrator uses those rates to estimate the state at later times.

The basic idea is:

```text
change in state ≈ rate of change × elapsed time
```

For example, a current slope of 600 A/s maintained for 0.001 s would add 0.6 A. In our motor, the slope changes as current and speed change, so a solver repeatedly reevaluates the equations.

We'll use SciPy's `solve_ivp` to integrate the two equations. It adjusts its internal step sizes to control numerical error. The times requested for plotting are output samples; they are not necessarily the internal integration steps.

The electrical timescale `L / R` is **0.01 s** for our parameters. Current can change much faster than the motor reaches its final speed, so the early part of the response deserves a closer look.

## 1.6 Check the simulation against the equations

A plausible-looking plot is useful, but we also need a numerical check. At steady state, both derivatives are zero. With zero external load, the equations reduce to:

```text
voltage = R · steady_current + K_e · steady_speed
K_t · steady_current = b · steady_speed
```

The second equation gives `steady_current = b · steady_speed / K_t`. Substituting it into the first and rearranging gives:

```text
steady_speed   = K_t · voltage / (R · b + K_t · K_e)
steady_current = b · steady_speed / K_t
```

For our parameters and 6 V input, the predicted values are approximately:

| Quantity | Steady-state prediction |
| --- | --- |
| Angular velocity | 54.55 rad/s, or about 521 rpm |
| Current | 0.5455 A |
| Motor torque | 0.05455 N·m |
| Back EMF | 5.455 V |

The resistive voltage drop is about 0.5455 V; together with back EMF, it accounts for the 6 V supply. Motor torque balances the viscous friction torque.

These are limiting values as time becomes large. The final sample of a finite simulation should be close, not necessarily identical. If it is still changing appreciably, extend the simulation before judging the steady-state comparison.

## What we've built

Once the exercises are complete, we have a model that converts voltage into current and motion, plus plots and steady-state checks that help us understand its behavior.

The key connection is that **current creates torque, net torque changes speed, and speed creates back EMF that affects current**. Neither current nor speed can be simulated independently of the other.
