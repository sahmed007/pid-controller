````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

R, L, K_t, K_e, J, b = 1.0, 0.01, 0.1, 0.1, 0.01, 0.001

def motor_dynamics(t, state, voltage, load_torque):
    current, omega = state
    return [(voltage - R*current - K_e*omega)/L,
            (K_t*current - b*omega - load_torque)/J]
class PID:
    def __init__(self, kp, ki, kd):
        self.kp, self.ki, self.kd = kp, ki, kd
        self.reset()
    def reset(self):
        self.integral = 0.0
        self.previous_error = None
        self.p_term = self.i_term = self.d_term = 0.0
    def update(self, setpoint, measurement, dt):
        if not np.isfinite(dt) or dt <= 0:
            raise ValueError("dt must be finite and positive")
        error = setpoint-measurement
        slope = 0.0 if self.previous_error is None else (error-self.previous_error)/dt
        self.integral += error*dt
        self.previous_error = error
        self.p_term = self.kp*error
        self.i_term = self.ki*self.integral
        self.d_term = self.kd*slope
        return self.p_term+self.i_term+self.d_term
def simulate(controller, target=lambda t: 40.0, load=lambda t: 0.0,
             duration=8.0, dt=0.001, noise=0.0, seed=42):
    # Each row describes the beginning of one control interval.
    count = round(duration/dt)
    if dt <= 0 or not np.isclose(count*dt, duration):
        raise ValueError("duration must be a positive integer multiple of dt")
    rng = np.random.default_rng(seed)
    state = np.array([0.0, 0.0])
    records = []
    for k in range(count):
        t = k*dt
        reference = target(t)
        measured = state[1] + noise*rng.normal()
        voltage = controller.update(reference, measured, dt)
        requested = getattr(controller, "requested", voltage)
        records.append([t, reference, state[0], state[1], measured,
                        requested, voltage, controller.p_term,
                        controller.i_term, controller.d_term])
        result = solve_ivp(motor_dynamics, (t, t+dt), state,
                          args=(voltage, load(t)), rtol=1e-7, atol=1e-9)
        if not result.success:
            raise RuntimeError(result.message)
        state = result.y[:, -1]
    keys = "time target current speed measured requested voltage p i d".split()
    return dict(zip(keys, np.asarray(records).T))

def plot_run(run, title):
    fig, axes = plt.subplots(3, 1, sharex=True, figsize=(10, 8), layout="constrained")
    axes[0].plot(run['time'], run['target'], '--', label='Target')
    axes[0].plot(run['time'], run['speed'], label='True speed')
    axes[0].set_ylabel('Speed (rad/s)')
    axes[1].plot(run['time'], run['requested'], label='Requested')
    axes[1].plot(run['time'], run['voltage'], '--', label='Applied')
    axes[1].set_ylabel('Voltage (V)')
    for key in ['p', 'i', 'd']:
        axes[2].plot(run['time'], run[key], label=key.upper())
    axes[2].set_ylabel('Contribution (V)')
    axes[2].set_xlabel('Time (s)')
    for ax in axes:
        ax.grid(alpha=.25); ax.legend()
    fig.suptitle(title)
    plt.show()
````

## 9.1 Respect actuator limits

Our drive can apply voltages between −6 V and +6 V. This assumes a bidirectional driver capable of braking and reverse voltage, not a bare unidirectional power supply.

$$V_{applied}=\operatorname{clip}(V_{requested},-6,6).$$

The motor responds to applied voltage. The controller can still request a much larger value. At 6 V and zero external load, our equilibrium speed is only 54.545 rad/s; a 100 rad/s target is therefore unreachable.

```python
class SaturatedPID(PID):
    def update(self,setpoint,measurement,dt):
        self.requested=super().update(setpoint,measurement,dt)
        return float(np.clip(self.requested,-6,6))
unreachable=simulate(SaturatedPID(.4,.5,0),target=lambda t:100.0,duration=4)
assert np.max(np.abs(unreachable['voltage']))<=6
plot_run(unreachable,'Unreachable target and voltage saturation')
```

Clipping changes the plant input but does not stop the old integral from accumulating. Large requested voltage is evidence of windup, not additional motor effort.

## 9.2 Prevent windup with conditional integration

When saturation prevents correction, persistent error can grow the integral far beyond useful values. After the target becomes reachable again, that stored command can keep the motor pushing in the wrong direction.

Conditional integration evaluates a candidate integral update. Reject it only when the candidate command exceeds a bound and the integral increment pushes farther beyond that bound. Allow updates that move back toward the allowed range. Use the applied-voltage limit in this decision.

For the derivative filter, a backward-Euler discretization of a first-order low-pass filter gives:

$$\alpha=\frac{\Delta t}{\tau_f+\Delta t},\quad d_{f,k}=d_{f,k-1}+\alpha(d_{raw,k}-d_{f,k-1}).$$

Setting $\tau_f=0$ makes $\alpha=1$, recovering the unfiltered derivative. Derivative-on-measurement uses $d_{raw}=-(\omega_k-\omega_{k-1})/\Delta t$ so a reference jump does not enter the derivative path.

```python
class PracticalPID(PID):
    def __init__(self, kp, ki, kd, limits=None, anti_windup=False,
                 derivative_mode="error", filter_tau=0.0):
        if derivative_mode not in ("error", "measurement"):
            raise ValueError("Unknown derivative mode")
        if not np.isfinite(filter_tau) or filter_tau < 0:
            raise ValueError("filter_tau must be finite and nonnegative")
        if limits is not None and not (np.isfinite(limits).all() and limits[0] < limits[1]):
            raise ValueError("limits must be an ordered pair of finite bounds")
        self.limits, self.anti_windup = limits, anti_windup
        self.derivative_mode, self.filter_tau = derivative_mode, filter_tau
        super().__init__(kp, ki, kd)
    def reset(self):
        super().reset()
        self.previous_measurement = None
        self.filtered_slope = 0.0
        self.requested = 0.0
    def update(self, setpoint, measurement, dt):
        if not np.isfinite(dt) or dt <= 0:
            raise ValueError("dt must be finite and positive")
        error = setpoint-measurement
        if self.previous_error is None:
            slope = 0.0
        elif self.derivative_mode == "measurement":
            slope = -(measurement-self.previous_measurement)/dt
        else:
            slope = (error-self.previous_error)/dt
        alpha = dt/(self.filter_tau+dt)
        self.filtered_slope += alpha*(slope-self.filtered_slope)
        self.p_term = self.kp*error
        self.d_term = self.kd*self.filtered_slope
        candidate = self.integral+error*dt
        candidate_output = self.p_term+self.ki*candidate+self.d_term
        increment = self.ki*(candidate-self.integral)
        pushing = self.limits is not None and (
            (candidate_output>self.limits[1] and increment>0) or
            (candidate_output<self.limits[0] and increment<0))
        if not (self.anti_windup and pushing):
            self.integral = candidate
        self.i_term = self.ki*self.integral
        self.requested = self.p_term+self.i_term+self.d_term
        self.previous_error, self.previous_measurement = error, measurement
        return self.requested if self.limits is None else float(np.clip(self.requested,*self.limits))
```

This implementation uses conditional integration, not back-calculation. It prevents additional harmful accumulation but does not instantly erase arbitrary pre-existing integral state. The first derivative sample remains zero.

</details>

```python
schedule=lambda t:100.0 if t<4 else 30.0
configs={'ideal':PracticalPID(.4,.5,0),
         'saturated':PracticalPID(.4,.5,0,limits=(-6,6)),
         'protected':PracticalPID(.4,.5,0,limits=(-6,6),anti_windup=True)}
recovery_runs={name:simulate(c,target=schedule,duration=12) for name,c in configs.items()}
def recovery_time(run):
    t,y=run['time'],run['speed']
    outside=np.flatnonzero((t>=4)&(np.abs(y-30)>.6))
    if len(outside)==0:return 0.0
    return np.inf if outside[-1]==len(t)-1 else float(t[outside[-1]+1]-4)
for name,trial in recovery_runs.items():
    print(name,'recovery seconds:',recovery_time(trial))
    if name!='ideal':assert np.max(np.abs(trial['voltage']))<=6
    plt.plot(trial['time'],trial['speed'],label=name)
plt.plot(trial['time'],trial['target'],'--',label='Target')
plt.xlabel('Time (s)');plt.ylabel('Speed (rad/s)');plt.legend();plt.show()
assert recovery_time(recovery_runs['protected'])<recovery_time(recovery_runs['saturated'])
plot_run(recovery_runs['protected'],'Protected controller recovery')
```

The ideal run is a comparison of assumptions, not a physically achievable benchmark under the ±6 V limit.

</details>

## 9.4 Remove derivative kick and filter noise

Derivative-on-measurement removes the direct setpoint derivative. It does not remove the proportional step or the actual motor response. Filtering reduces high-frequency noise at the cost of lag: the derivative then reacts to a smoothed estimate, which can change damping.

Use a filter time constant of 0.02 s with a 0.001 s sample interval, giving $\alpha\approx0.0476$. Compare equal noise sequences using the same seed.

<div class="lesson-callout" style="border-left: 3px solid #b99a62; background-color: rgba(185, 154, 98, 0.08); padding: 12px 18px; margin: 16px 0;">

**Key distinction**

Filtering cannot turn a noisy sensor into perfect information. Compare true speed, measured speed, and command effort; quieter commands alone do not establish better tracking.

</div>

## What the experiment establishes

The protected saturated controller recovers to the lower target within about 2.76 seconds in this experiment; the unprotected one has not recovered by the end of the observation. Filtering reduces derivative RMS from about 2.83 to 0.094 V for the seeded noise experiment. These results illustrate specific mechanisms, not a universal tuning guarantee. You now have a complete simulated path from motor dynamics to a practical sampled PID controller.
